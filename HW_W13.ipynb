## Concept Drift

<div dir=rtl>
**Concept Drift** به پدیده‌ای گفته می‌شود که در آن توزیع داده‌ها یا رابطه بین ورودی‌ها و خروجی‌ها در طول زمان تغییر می‌کند و باعث می‌شود مدلی که قبلاً آموزش دیده دیگر عملکرد مطلوبی نداشته باشد. در یادگیری ماشین کلاسیک فرض می‌شود داده‌های آموزش و آینده از یک توزیع ثابت می‌آیند، اما در محیط‌های واقعی این فرض اغلب نقض می‌شود. این تغییر می‌تواند در توزیع ویژگی‌ها `P(X)`، در توزیع کلاس‌ها `P(Y)`، یا در رابطه شرطی بین آن‌ها `P(Y | X)` رخ دهد که نوع سوم مهم‌ترین و پیچیده‌ترین حالت است. Concept Drift معمولاً در سیستم‌هایی دیده می‌شود که داده‌ها به‌صورت جریان پیوسته (Data Stream) وارد می‌شوند و محیط پویا است. از مهم‌ترین حوزه‌های مطالعه آن می‌توان به تشخیص تقلب مالی، بازارهای مالی، امنیت سایبری، سیستم‌های IoT صنعتی، سیستم‌های توصیه‌گر و حوزه سلامت اشاره کرد. در این کاربردها رفتار کاربران، بازار یا مهاجمان به‌طور مداوم تغییر می‌کند و مدل باید بتواند خود را با شرایط جدید تطبیق دهد. اگر Drift نادیده گرفته شود، دقت مدل کاهش یافته و تصمیم‌های نادرست افزایش می‌یابد، بنابراین استفاده از روش‌های یادگیری آنلاین و مدل‌های تطبیقی برای مدیریت این پدیده ضروری است.

## Ensemble

<div dir=rtl>
مدل‌های **Ensemble** به رویکردی در یادگیری ماشین گفته می‌شود که در آن به‌جای استفاده از یک مدل منفرد، چندین مدل به‌صورت همزمان آموزش داده شده و خروجی آن‌ها با یکدیگر ترکیب می‌شود تا تصمیم نهایی تولید شود. ایده اصلی این است که ترکیب چند پیش‌بینی معمولاً پایدارتر و دقیق‌تر از یک پیش‌بینی واحد است، زیرا خطاهای مدل‌ها تا حدی یکدیگر را خنثی می‌کنند. روش‌هایی مانند Bagging و Boosting نمونه‌های رایج Ensemble هستند که یا مدل‌ها را به‌صورت مستقل آموزش می‌دهند یا به‌صورت ترتیبی خطاهای قبلی را اصلاح می‌کنند. در مسئله **Concept Drift**، مدل‌های Ensemble اهمیت ویژه‌ای دارند زیرا می‌توان مدل‌های جدید را برای داده‌های جدید اضافه کرد و به مدل‌های قدیمی وزن کمتری داد یا آن‌ها را حذف کرد؛ در نتیجه سیستم به‌صورت تطبیقی با تغییر توزیع داده‌ها سازگار می‌شود. این رویکرد به‌جای بازنویسی کامل یک مدل، حافظه را میان چند مدل توزیع می‌کند. ارتباط این موضوع با **Catastrophic Forgetting** در این است که در یادگیری پیوسته، اگر یک مدل منفرد فقط با داده‌های جدید به‌روزرسانی شود، ممکن است دانش قبلی خود را به‌شدت از دست بدهد. اما در Ensemble، چون مدل‌های قدیمی همچنان بخشی از ساختار تصمیم‌گیری باقی می‌مانند، خطر فراموشی کامل دانش گذشته کاهش می‌یابد. به بیان دیگر، Ensemble نوعی مکانیزم حافظه توزیع‌شده ایجاد می‌کند که در برابر فراموشی ناگهانی مقاوم‌تر است. با این حال، اگر وزن‌دهی مدل‌ها به‌درستی طراحی نشود یا مدل‌های قدیمی خیلی سریع کنار گذاشته شوند، همچنان امکان بروز مشکلات حافظه وجود دارد. بنابراین در طراحی سیستم‌های تطبیقی، باید تعادل میان یادگیری اطلاعات جدید و حفظ دانش قدیمی به‌دقت مدیریت شود.

## Imbalance

<div dir=rtl>
عدم توازن داده‌ها (**Imbalanced Data**) زمانی رخ می‌دهد که تعداد نمونه‌های یک کلاس به‌طور قابل توجهی بیشتر از کلاس دیگر باشد، به‌گونه‌ای که توزیع کلاس‌ها نامتقارن شود؛ برای مثال اگر در یک مسئله دودویی داشته باشیم `P(Y=1) ≪ P(Y=0)`، داده‌ها نامتوازن محسوب می‌شوند. این وضعیت معمولاً در مسائل تشخیص تقلب، تشخیص بیماری‌های نادر، کشف نفوذ سایبری و خطای صنعتی دیده می‌شود، زیرا رخدادهای مورد نظر ذاتاً کم‌تعداد هستند. مشکل اصلی این است که مدل ممکن است با پیش‌بینی همیشگی کلاس اکثریت به دقت ظاهراً بالا برسد، اما عملکرد بسیار ضعیفی روی کلاس اقلیت داشته باشد. برای مقابله با این مشکل چند راهکار پیشنهاد شده است: نخست روش‌های سطح داده (Data-Level) مانند Oversampling برای افزایش نمونه‌های کلاس اقلیت یا Undersampling برای کاهش نمونه‌های کلاس اکثریت؛ دوم روش‌های سطح الگوریتم (Algorithm-Level) مانند تنظیم وزن کلاس‌ها (Class Weighting) یا استفاده از Cost-Sensitive Learning که جریمه خطای کلاس اقلیت را بیشتر می‌کند؛ سوم استفاده از توابع زیان خاص مانند Focal Loss که تمرکز بیشتری بر نمونه‌های سخت و اقلیت دارد؛ و چهارم انتخاب معیارهای ارزیابی مناسب مانند Precision، Recall، F1-Score و AUC به‌جای Accuracy. در نهایت، انتخاب راهکار مناسب به نوع مسئله، میزان عدم توازن و حساسیت کاربرد به خطای کلاس اقلیت بستگی دارد.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

# -----------------------------
# 1) Reproducibility (تکرارپذیری)
# -----------------------------
np.random.seed(42)

# -----------------------------
# 2) Define imbalance sizes
# -----------------------------
n0 = 900  # class 0 (majority)
n1 = 100  # class 1 (minority)

# -----------------------------
# 3) Define feature distributions (partially separable)
# -----------------------------
mu0, sigma0 = 0.0, 1.0
mu1, sigma1 = 2.0, 1.0

# -----------------------------
# 4) Sample the feature values (1 feature => shape (n,1))
# -----------------------------
X0 = np.random.normal(loc=mu0, scale=sigma0, size=(n0, 1))
X1 = np.random.normal(loc=mu1, scale=sigma1, size=(n1, 1))

# -----------------------------
# 5) Create labels
# -----------------------------
y0 = np.zeros(n0, dtype=int)
y1 = np.ones(n1, dtype=int)

# -----------------------------
# 6) Combine
# -----------------------------
X = np.vstack([X0, X1])          # (1000, 1)
y = np.concatenate([y0, y1])     # (1000,)

# -----------------------------
# 7) Shuffle (recommended)
# -----------------------------
perm = np.random.permutation(len(y))
X = X[perm]
y = y[perm]

# -----------------------------
# 8) Train/Test split (common: test_size=0.30)
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,      # 30% test (رایج)
    random_state=42,
    stratify=y
)

# -----------------------------
# 9) Sanity checks
# -----------------------------
def class_counts(arr):
    unique, counts = np.unique(arr, return_counts=True)
    return dict(zip(unique, counts))

print("Total:", class_counts(y))
print("Train:", class_counts(y_train), " -> size:", len(y_train))
print("Test :", class_counts(y_test),  " -> size:", len(y_test))

print("\nClass-1 ratio (Total):", y.mean())
print("Class-1 ratio (Train):", y_train.mean())
print("Class-1 ratio (Test) :", y_test.mean())

# -----------------------------
# 10) Visualization: show overlap (همپوشانی)
# -----------------------------
x0 = X[y == 0].ravel()
x1 = X[y == 1].ravel()

plt.figure(figsize=(8, 5))
plt.hist(x0, bins=50, density=True, alpha=0.5, label="Class 0 (900 samples)")
plt.hist(x1, bins=50, density=True, alpha=0.5, label="Class 1 (100 samples)")
plt.xlabel("Feature value (x)")
plt.ylabel("Density")
plt.title("Feature distribution overlap (Histogram)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Optional: scatter with jitter (برای دیدن نمونه‌ها)
rng = np.random.RandomState(42)
j0 = rng.normal(loc=0.0, scale=0.03, size=len(x0))
j1 = rng.normal(loc=1.0, scale=0.03, size=len(x1))

plt.figure(figsize=(8, 3))
plt.scatter(x0, j0, alpha=0.25, s=12, label="Class 0")
plt.scatter(x1, j1, alpha=0.8,  s=18, label="Class 1")
plt.yticks([0, 1], ["y=0", "y=1"])
plt.xlabel("Feature value (x)")
plt.title("Sample-level view (jitter plot) — overlap is visible on x-axis")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, ConfusionMatrixDisplay
)

# =============================
# Part 1) Data generation (Imbalanced + partially separable)
# =============================
np.random.seed(42)

n0, n1 = 900, 100
mu0, sigma0 = 0.0, 1.0
mu1, sigma1 = 2.0, 1.0

X0 = np.random.normal(loc=mu0, scale=sigma0, size=(n0, 1))
X1 = np.random.normal(loc=mu1, scale=sigma1, size=(n1, 1))

y0 = np.zeros(n0, dtype=int)
y1 = np.ones(n1, dtype=int)

X = np.vstack([X0, X1])
y = np.concatenate([y0, y1])

perm = np.random.permutation(len(y))
X = X[perm]
y = y[perm]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# =============================
# Part 2) Train a baseline model: Logistic Regression
# =============================

# (Recommended) scaling for logistic regression (especially if features differ in scale)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

model = LogisticRegression(
    solver="lbfgs",
    max_iter=1000,
    random_state=42
)

model.fit(X_train_scaled, y_train)

# Default threshold = 0.5
y_pred = model.predict(X_test_scaled)

# =============================
# Part 2.2) Compute metrics on test
# =============================
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, zero_division=0)
rec = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

cm = confusion_matrix(y_test, y_pred)  # [[TN, FP], [FN, TP]]
TN, FP, FN, TP = cm.ravel()

print("=== Test Metrics (threshold=0.5) ===")
print(f"Accuracy  : {acc:.4f}")
print(f"Precision : {prec:.4f}")
print(f"Recall    : {rec:.4f}")
print(f"F1-score  : {f1:.4f}")

print("\n=== Confusion Matrix ===")
print(cm)
print(f"\nTN={TN}, FP={FP}, FN={FN}, TP={TP}")

# =============================
# Part 2.3) Interpretation helpers (Decision-focused)
# =============================
# Rates that are often more interpretable
fnr = FN / (FN + TP) if (FN + TP) > 0 else 0.0  # False Negative Rate = 1 - Recall
fpr = FP / (FP + TN) if (FP + TN) > 0 else 0.0  # False Positive Rate

print("\n=== Error Rates ===")
print(f"FNR (miss rate) = FN / (FN + TP) = {fnr:.4f}")
print(f"FPR (false alarm rate) = FP / (FP + TN) = {fpr:.4f}")

# =============================
# Optional: plot confusion matrix
# =============================
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[0, 1])
disp.plot(values_format="d")
plt.title("Confusion Matrix (Logistic Regression, threshold=0.5)")
plt.grid(False)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score, confusion_matrix

# -----------------------------
# 1) Get probabilities for class 1 (positive class)
# -----------------------------
X_test_scaled = scaler.transform(X_test)
proba_pos = model.predict_proba(X_test_scaled)[:, 1]  # P(y=1|x)

# -----------------------------
# 2) Evaluate multiple thresholds
# -----------------------------
thresholds = np.arange(0.1, 1.0, 0.1)  # 0.1 ... 0.9

precisions = []
recalls = []
fps = []
fns = []

print("thr | precision | recall | FP | FN")
print("-" * 38)

for t in thresholds:
    # Manual prediction based on threshold
    y_pred_t = (proba_pos >= t).astype(int)

    # Metrics
    p = precision_score(y_test, y_pred_t, zero_division=0)
    r = recall_score(y_test, y_pred_t, zero_division=0)

    cm = confusion_matrix(y_test, y_pred_t)
    TN, FP, FN, TP = cm.ravel()

    precisions.append(p)
    recalls.append(r)
    fps.append(FP)
    fns.append(FN)

    print(f"{t:0.1f} |   {p:0.4f}  | {r:0.4f} | {FP:2d} | {FN:2d}")

# -----------------------------
# 3) Plot Precision-Recall tradeoff (threshold sweep)
# -----------------------------
plt.figure(figsize=(7, 5))
plt.plot(recalls, precisions, marker="o")
plt.xlabel("Recall (TP / (TP + FN))")
plt.ylabel("Precision (TP / (TP + FP))")
plt.title("Precision–Recall Tradeoff (varying threshold)")
plt.grid(True, alpha=0.3)

# Annotate points with thresholds (for decision making)
for i, t in enumerate(thresholds):
    plt.annotate(f"{t:0.1f}", (recalls[i], precisions[i]), textcoords="offset points", xytext=(6, 4))

plt.show()

# -----------------------------
# 4) Optional: show FN/FP vs threshold (decision-cost view)
# -----------------------------
plt.figure(figsize=(7, 5))
plt.plot(thresholds, fns, marker="o", label="FN")
plt.plot(thresholds, fps, marker="o", label="FP")
plt.xlabel("Threshold")
plt.ylabel("Count")
plt.title("FN and FP vs Threshold")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import roc_curve, roc_auc_score

# =============================
# Part 1) Generate imbalanced + partially separable data
# =============================
np.random.seed(42)

n0, n1 = 900, 100
mu0, sigma0 = 0.0, 1.0
mu1, sigma1 = 2.0, 1.0

X0 = np.random.normal(loc=mu0, scale=sigma0, size=(n0, 1))
X1 = np.random.normal(loc=mu1, scale=sigma1, size=(n1, 1))

y0 = np.zeros(n0, dtype=int)
y1 = np.ones(n1, dtype=int)

X = np.vstack([X0, X1])
y = np.concatenate([y0, y1])

perm = np.random.permutation(len(y))
X = X[perm]
y = y[perm]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# =============================
# Part 2) Train Logistic Regression
# =============================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression(solver="lbfgs", max_iter=1000, random_state=42)
model.fit(X_train_scaled, y_train)

# =============================
# Part 3) ROC + AUC needs scores/probabilities (NOT hard labels)
# =============================
# Take probability for positive class: P(y=1|x)
y_score = model.predict_proba(X_test_scaled)[:, 1]

# ROC curve points
fpr, tpr, thresholds = roc_curve(y_test, y_score)

# AUC value
auc = roc_auc_score(y_test, y_score)

print(f"AUC = {auc:.4f}")

# =============================
# Part 4) Plot ROC curve
# =============================
plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, linewidth=2, label=f"LogReg (AUC={auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--", linewidth=2, label="Random (AUC=0.5)")
plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR = Recall)")
plt.title("ROC Curve")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix,
    recall_score,
    balanced_accuracy_score,
    roc_auc_score
)

from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE

np.random.seed(42)

# -------------------------
# 1) Generate imbalanced dataset
# -------------------------
X, y = make_classification(
    n_samples=2000,
    n_features=5,
    n_informative=3,
    n_redundant=0,
    weights=[0.9, 0.1],
    class_sep=1.5,
    random_state=42
)

# -------------------------
# 2) Train-test split
# -------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    stratify=y,
    random_state=42
)

# -------------------------
# 3) Apply SMOTE ONLY on train
# -------------------------
smote = SMOTE(random_state=42)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)

print("Before SMOTE:", np.bincount(y_train))
print("After SMOTE:", np.bincount(y_train_sm))

# -------------------------
# 4) Train model again
# -------------------------
model_sm = RandomForestClassifier(random_state=42)
model_sm.fit(X_train_sm, y_train_sm)

# -------------------------
# 5) Evaluate
# -------------------------
y_pred_sm = model_sm.predict(X_test)
y_proba_sm = model_sm.predict_proba(X_test)[:, 1]

cm_sm = confusion_matrix(y_test, y_pred_sm)
recall_sm = recall_score(y_test, y_pred_sm)
bal_acc_sm = balanced_accuracy_score(y_test, y_pred_sm)
auc_sm = roc_auc_score(y_test, y_proba_sm)

print("\n=== After SMOTE ===")
print("Confusion Matrix:\n", cm_sm)
print("Recall (positive class):", recall_sm)
print("Balanced Accuracy:", bal_acc_sm)
print("ROC AUC:", auc_sm)
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, recall_score, precision_score

# -----------------------------------------
# A) Evaluate one threshold (prints CM + metrics)
# -----------------------------------------
def evaluate_threshold(y_true, y_proba, thr):
    y_pred_thr = (y_proba >= thr).astype(int)
    
    cm = confusion_matrix(y_true, y_pred_thr)
    recall = recall_score(y_true, y_pred_thr)
    precision = precision_score(y_true, y_pred_thr, zero_division=0)

    TN, FP, FN, TP = cm.ravel()

    print(f"\nThreshold = {thr:.2f}")
    print("Confusion Matrix:\n", cm)
    print(f"TP={TP}, FP={FP}, FN={FN}, TN={TN}")
    print(f"Recall    = {recall:.4f}  (TP / (TP+FN))")
    print(f"Precision = {precision:.4f} (TP / (TP+FP))")
    
    return thr, recall, precision, cm

# -----------------------------------------
# B) Show how Recall increases when threshold decreases
# -----------------------------------------
thresholds = [0.80, 0.70, 0.60, 0.50, 0.40, 0.30, 0.20]

print("\n=== Threshold sweep (see Recall vs Precision trade-off) ===")
sweep_results = []
for thr in thresholds:
    sweep_results.append(evaluate_threshold(y_test, y_proba_sm, thr))

# -----------------------------------------
# C) Control Precision: choose threshold with Precision >= target
#     and among them maximize Recall
# -----------------------------------------
target_precision = 0.90

valid = [r for r in sweep_results if r[2] >= target_precision]  # r[2] = precision
if len(valid) == 0:
    print(f"\nNo threshold in the tested set keeps Precision >= {target_precision:.2f}")
else:
    best = max(valid, key=lambda x: x[1])  # maximize recall among valid
    best_thr, best_recall, best_precision, best_cm = best

    print("\n=== Best threshold under Precision control ===")
    print(f"Target Precision >= {target_precision:.2f}")
    print(f"Chosen Threshold = {best_thr:.2f}")
    print("Confusion Matrix:\n", best_cm)
    print(f"Recall    = {best_recall:.4f}")
    print(f"Precision = {best_precision:.4f}")
```

```python
import numpy as np
from sklearn.metrics import (
    confusion_matrix,
    recall_score,
    precision_score,
    balanced_accuracy_score,
    roc_auc_score
)

# -------------------------
# Helper: evaluate model on a given test set
# -------------------------
def evaluate_model(model, X_eval, y_eval, threshold=0.5, title=""):
    # Probabilities for the positive class
    y_proba = model.predict_proba(X_eval)[:, 1]
    # Thresholded predictions
    y_pred = (y_proba >= threshold).astype(int)

    cm = confusion_matrix(y_eval, y_pred)
    TN, FP, FN, TP = cm.ravel()

    recall = recall_score(y_eval, y_pred)
    precision = precision_score(y_eval, y_pred, zero_division=0)
    bal_acc = balanced_accuracy_score(y_eval, y_pred)
    auc = roc_auc_score(y_eval, y_proba)

    print(f"\n=== {title} (threshold={threshold}) ===")
    print("Confusion Matrix:\n", cm)
    print(f"TP={TP}, FP={FP}, FN={FN}, TN={TN}")
    print(f"Recall    = {recall:.4f}  (TP / (TP+FN))")
    print(f"Precision = {precision:.4f} (TP / (TP+FP))")
    print(f"Balanced Accuracy = {bal_acc:.4f}")
    print(f"ROC AUC = {auc:.4f}")

    return {"cm": cm, "recall": recall, "precision": precision, "bal_acc": bal_acc, "auc": auc}

# -------------------------
# 1) Baseline evaluation on original test
# -------------------------
base = evaluate_model(
    model_sm, X_test, y_test,
    threshold=0.5,
    title="Original TEST (no drift)"
)

# -------------------------
# 2) Simulate FEATURE DRIFT on test (mean shift OR variance change)
#    We'll modify ONE feature (e.g., feature index 0) in X_test only.
# -------------------------
X_test_drift = X_test.copy()

feature_idx = 0

# Option A: Mean shift (add constant)
mean_shift = 0.8
X_test_drift[:, feature_idx] = X_test_drift[:, feature_idx] + mean_shift

# Option B: Variance change (scale)
# var_scale = 1.4
# X_test_drift[:, feature_idx] = X_test_drift[:, feature_idx] * var_scale

# -------------------------
# 3) Evaluate on drifted test WITHOUT retraining
# -------------------------
drifted = evaluate_model(
    model_sm, X_test_drift, y_test,
    threshold=0.5,
    title=f"DRIFTED TEST (feature {feature_idx} mean shift = +{mean_shift})"
)

# -------------------------
# 4) (Optional) Compare differences numerically
# -------------------------
print("\n=== Metric deltas (drift - base) ===")
print(f"ΔRecall    = {drifted['recall'] - base['recall']:+.4f}")
print(f"ΔPrecision = {drifted['precision'] - base['precision']:+.4f}")
print(f"ΔBalAcc    = {drifted['bal_acc'] - base['bal_acc']:+.4f}")
print(f"ΔAUC       = {drifted['auc'] - base['auc']:+.4f}")
```
